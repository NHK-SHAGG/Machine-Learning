Bài 3. Hãy chọn 1 bộ dữ liệu tùy ý (HD: có thể tham khảo trên các trang mạng, ví dụ như
Kaggle) và thực hiện các bước cần thiết để có được mô hình hồi qui ứng với bộ dữ liệu đã
chọn.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
X = np.array([[30], [35], [45], [50], [60], [65], [75], [85], [90], [100], [110], [120], [130], [140], [150]])
y = np.array([1.45, 1.700, 2.100, 2.400, 3.000, 3.150, 3.700, 4.150, 4.400, 5.000, 5.300, 5.900, 6.200, 6.800, 7.200])
model = LinearRegression()
model.fit(X, y)

print("--- Thông số mô hình học được từ 15 căn nhà ---")
print(f"Hệ số góc (a): {model.coef_[0]:.4f}")
print(f"Hệ số chắn (b): {model.intercept_:.4f}")
print(f"Phương trình xu hướng: y = {model.coef_[0]:.4f} * x + {model.intercept_:.4f}")
print(f"Với x là diện tích và y là giá nhà")

dien_tich_can_du_doan = np.array([[95], [165]])
gia_du_doan = model.predict(dien_tich_can_du_doan)

print(f"\n--- Kết quả dự đoán ---")
print(f"Dự đoán giá nhà 95 m²: {gia_du_doan[0]:,.4f} tỉ VNĐ")
print(f"Dự đoán giá nhà 165 m²: {gia_du_doan[1]:,.4f} tỉ VNĐ")

X_dai = np.linspace(25, 170, 100).reshape(-1, 1)
y_dai_pred = model.predict(X_dai)
plt.figure(figsize=(10, 6))
plt.scatter(X, y, color='blue', label='Dữ liệu thực tế (15 căn nhà)', s=50)
plt.plot(X_dai, y_dai_pred, color='red', linewidth=2, label='Đường xu hướng hồi quy tuyến tính')
plt.scatter(dien_tich_can_du_doan, gia_du_doan, color='green', marker='X', s=150, label='Điểm dự đoán (95m² & 165m²)')
plt.xlabel('Diện tích ($m^2$)')
plt.ylabel('Giá nhà (tỉ VNĐ)')
plt.title('DỰ ĐOÁN GIÁ NHÀ')
plt.legend()
plt.grid(True)
plt.show()
```
